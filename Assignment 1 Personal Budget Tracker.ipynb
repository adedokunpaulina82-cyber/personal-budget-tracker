```python
# Store one month's expenses

expenses = [
    {"category": "Food", "item": "Rice", "amount": 12500},
    {"category": "Food", "item": "Beans", "amount": 7000},
    {"category": "Food", "item": "Indomie", "amount": 5000},
    {"category": "Transport", "item": "T-fare", "amount": 4000},
    {"category": "Transport", "item": "Fuel", "amount": 2500},
    {"category": "Airtime", "item": "Phone credit", "amount": 4500},
    {"category": "Rent", "item": "House rent", "amount": 30000},
    {"category": "Food", "item": "Eggs", "amount": 2000},
    {"category": "Subscription", "item": "Light", "amount": 17000}
]

# Set the budget for each category

budgets = {
    "Food": 20000,
    "Transport": 8000,
    "Airtime": 5000,
    "Rent": 30000,
    "Subscription": 15000
}
```

```python
# Create an empty dictionary to store totals

totals = {}

# Loop through each expense
for e in expenses:
    category = e["category"]
    amount = e["amount"]

    # Add the amount to the category total
    if category in totals:
        totals[category] += amount
    else:
        totals[category] = amount

print(totals)
```

```python

# Compare each category's spending with its budget

for category in budgets:
    spent = totals[category]
    budget = budgets[category]

    # Check if spending is above the budget
    if spent > budget:
        difference = spent - budget
        status = f"OVER by ₦{difference:,.0f}"
    else:
        status = "OK"

    # Print the spending summary
    print(f"{category} spent ₦{spent:,.0f} of ₦{budget:,.0f} - {status}")
```
